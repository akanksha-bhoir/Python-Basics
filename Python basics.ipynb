**SET: A set is a collection which is unordered, unchangeable, and unindexed. Set items are unchangeable, but you can remove items and add new items.
The values True,0 and values False,1 are considered duplicate**

```python
_newset = {'abc', True,1, False,0, 'cream'}
print(_newset)
print(type(_newset))
```

```python
#there is no indexing in sets, so when you print the elements of set they will occur always in different order.
```

```python
print(len(_newset))
```

```python
print('cream' in _newset)
```

```python
if cream in _newset:
    print('Yes, Cream is present')
else:
    print('Not present')
```

```python
#Once a set is created, you cannot change its items, but you can add new items.

_newset.add('dream')
print(_newset)
```

```python
#To add items from another set into the current set, use the update() method.

new2={True,'happy happy', 'Marie Gold', 'Parle G'}
_newset.update(new2)
print(_newset)
```

```python
#To remove an item in a set, use the remove(), or the discard() method.
#If the item to remove does not exist, remove() will raise an error.

_newset.remove('give')
_newset.discard('give')
```

```python
#clear ,ethod empties the set but the empy set is present and #del method will completely delete the set
```

```python
for x in _newset:
    print(x)
```

**WHILE LOOP**

```python
i=0
while i<11:
    print(i)
    i+=1
```

```python
#With the break statement we can stop the loop even if the while condition is true:

i=0
while i<13:
    print(i)
    if i==7:
        break
    i+=1
```

```python
#With the continue statement we can stop the current iteration, and continue with the next:

i=0
while i<15:
    i+=1
    if i==9:
     continue
    print(i)
        
```

```python
i = 1
while i < 6:
  print(i)
  i += 1
else:
  print(f"{i} is no longer less than 6")
```
